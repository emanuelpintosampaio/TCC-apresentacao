## 3. Ver os slides

Clique nos slides e use as setas do teclado ou a barra de espaço para avançar. Para ver em tela cheia, use a última linha, que abre os slides numa aba nova.

```python
import subprocess
from google.colab import output

# Serve a pasta dos slides (o HTML e os vídeos) dentro do Colab
subprocess.Popen(["python3", "-m", "http.server", "8000"], cwd="/content/TCC-apresentacao/manim",
                 stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

output.serve_kernel_port_as_iframe(8000, path="/apresentacao.html", height=540)
# output.serve_kernel_port_as_window(8000, path="/apresentacao.html")   # abre numa aba nova
```
